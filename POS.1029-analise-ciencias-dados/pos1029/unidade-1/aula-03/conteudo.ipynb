### MÓDULO 1 - LISTAS

### 1. Listas (lists)

Uma lista é uma estrutura de dados que armazena uma sequência ordenada de elementos. Os elementos podem ser de qualquer tipo - números, textos, booleanos ou até outras listas.

### 1.1 Criando e Acessando Listas

Sintaxe básica: os elementos são separados por vírgulas e delimitados por colchetes.

````python
# Criando listas
frutas = ["maçã", "banana", "laranja", "uva"]
numeros = [10, 25, 37, 42, 58]
mista  = [1, "Python", True, 3.14]

# Acessando pelo índice (inicia em 0)
print(frutas[0])   # maçã

# Fatiamento (slicing)  [início : fim : passo]
print(frutas[1:3]) # ["banana", "laranja"]
print(numeros[::2])# [10, 37, 58]
````

Índice negativo: lista[-1] retorna o último elemento; lista[-2] o penúltimo, e assim por diante.

💡

### 1.2 Principais Métodos de Listas

Os métodos abaixo são os mais utilizados no dia a dia:

````python
lista = [3, 1, 4, 1, 5, 9]

lista.append(2)      # adiciona 2 ao final
lista.insert(0, 0)   # insere 0 na posição 0
lista.remove(1)      # remove a primeira ocorrência do valor 1
lista.pop()          # remove e retorna o último elemento
lista.pop(2)         # remove e retorna o índice 2
lista.sort()         # ordena in-place (menor → maior)
lista.reverse()      # inverte a ordem in-place
lista.index(5)       # retorna o índice de 5
lista.count(1)       # conta quantas vezes 1 aparece
len(lista)           # comprimento da lista
````

### 1.3 Iterando sobre Listas

````python
# for simples
for fruta in frutas:
    print(fruta)

# enumerate — índice + valor
for i, fruta in enumerate(frutas):
    print(f"{i}: {fruta}")

# list comprehension — cria lista nova
quadrados = [x**2 for x in range(1, 6)]
# [1, 4, 9, 16, 25]

# com filtro
pares = [x for x in range(10) if x % 2 == 0]
````

### 1.4 Exemplo Prático - Gerenciador de Notas

A seguir, um mini-programa que cadastra notas de alunos e calcula a média da turma:

````python
notas = []

# Entrada de dados
for i in range(5):
    nota = float(input(f"Digite a nota do aluno {i+1}: "))
    notas.append(nota)

# Cálculos
media    = sum(notas) / len(notas)
maior    = max(notas)
menor    = min(notas)
aprovados = [n for n in notas if n >= 6.0]

print(f"Notas:      {notas}")
print(f"Média:      {media:.2f}")
print(f"Maior nota: {maior}")
print(f"Menor nota: {menor}")
print(f"Aprovados:  {len(aprovados)} de {len(notas)}")
````

### MÓDULO 2 - DICIONÁRIOS

### 2. Dicionários (dicts)

Um dicionário armazena pares chave → valor. As chaves devem ser únicas e imutáveis (str, int, tuple); os valores podem ser de qualquer tipo.

### 2.1 Criando e Acessando Dicionários

````python
# Criação literal
aluno = {
    "nome":  "Ana Paula",
    "idade": 21,
    "curso": "Ciência da Computação",
    "notas": [8.5, 9.0, 7.5],
}

# Acesso
print(aluno["nome"])           # Ana Paula
print(aluno.get("idade"))      # 21
print(aluno.get("email", "—")) # — (padrão se ausente)

# Adicionando / atualizando
aluno["email"] = "ana@uni.br"
aluno["idade"] = 22

# Removendo
del aluno["email"]
matricula = aluno.pop("matricula", None) # seguro
````

### 2.2 Métodos Essenciais

````python
pessoa = {"nome": "Carlos", "cidade": "Fortaleza", "idade": 30}

pessoa.keys()    # dict_keys(["nome", "cidade", "idade"])
pessoa.values()  # dict_values(["Carlos", "Fortaleza", 30])
pessoa.items()   # dict_items([("nome", "Carlos"), ...])

# Verificar existência
"nome" in pessoa    # True
"email" in pessoa   # False

# Mesclar dicionários
extra = {"email": "carlos@mail.com"}
completo = pessoa | extra
````

### 2.3 Iterando sobre Dicionários

````python
# Percorrendo pares chave-valor
for chave, valor in aluno.items():
    print(f"{chave}: {valor}")

quadrados = {x: x**2 for x in range(1, 6)}
# {1: 1, 2: 4, 3: 9, 4: 16, 5: 25}

# Filtrar por condição
````

````python
maiores = {k: v for k, v in quadrados.items() if v > 9}
# {4: 16, 5: 25}
````

### 2.4 Exemplo Prático - Catálogo de Produtos

````python
catalogo = {
    "notebook": {"preco": 3200.00, "estoque": 15},
    "mouse":    {"preco":   89.90, "estoque": 40},
    "teclado":  {"preco":  150.00, "estoque":  8},
}

def exibir_catalogo(cat):
    print(f"{"Produto":<12} {"Preço":>10} {"Estoque":>9}")
    print("-" * 34)
    for produto, info in cat.items():
        print(f"{produto:<12} R${info["preco"]:>8.2f} {info["estoque"]:>9}")

exibir_catalogo(catalogo)

# Total em estoque
total = sum(i["preco"] * i["estoque"] for i in catalogo.values())
print(f"\nValor total em estoque: R$ {total:,.2f}")
````

### MÓDULO 3 - FUNÇÕES

### 3. Funções

Funções permitem encapsular um bloco de código reutilizável. Em Python elas são definidas com a palavra-chave

````python
def saudacao(nome):
    """Exibe uma mensagem de boas-vindas."""
    print(f"Olá, {nome}! Bem-vindo ao Python.")

saudacao("Maria")  # Olá, Maria! Bem-vindo ao Python.
````

💡 Docstring: Sempre documente suas funções com uma string entre aspas triplas logo após o def. Isso facilita a manutenção e gera documentação automática com help().

### 3.1 Parâmetros e Retorno

````python
# Parâmetros com valor padrão
def potencia(base, expoente=2):
    return base ** expoente

print(potencia(3))     # 9  (expoente padrão)
print(potencia(2, 10)) # 1024

# Retorno múltiplo (tupla)
def estatisticas(nums):
    return min(nums), max(nums), sum(nums) / len(nums)

mi, ma, med = estatisticas([4, 7, 2, 9, 1])
print(f"Min={mi}  Max={ma}  Média={med}")
````

### 3.2 *args e **kwargs

````python
# *args — número variável de argumentos posicionais
def somar(*numeros):
    return sum(numeros)

print(somar(1, 2, 3))        # 6
print(somar(10, 20, 30, 40)) # 100

# **kwargs — argumentos nomeados arbitrários
def apresentar(**dados):
    for chave, valor in dados.items():
        print(f"  {chave}: {valor}")

apresentar(nome="Lucas", cidade="Recife", profissao="Dev")
````

### 3.4 Exemplo Prático - Calculadora de IMC

````python
def calcular_imc(peso, altura):
    """Calcula o Índice de Massa Corporal."""
    return peso / altura ** 2

def classificar_imc(imc):
    """Retorna a classificação do IMC."""
    if imc < 18.5:
        return "Abaixo do peso"
    elif imc < 25.0:
        return "Peso normal"
    elif imc < 30.0:
        return "Sobrepeso"
    else:
        return "Obesidade"

def relatorio_imc(nome, peso, altura):
    """Gera relatório completo de IMC."""
    imc = calcular_imc(peso, altura)
    classe = classificar_imc(imc)
    print(f"Paciente : {nome}")
    print(f"IMC      : {imc:.2f}")
    print(f"Situação : {classe}")

relatorio_imc("João Silva", 82, 1.75)
````

### 4.2 Criando Arrays NumPy

````python
import numpy as np

# A partir de listas Python
v = np.array([1, 2, 3, 4, 5])          # 1-D
M = np.array([[1, 2, 3], [4, 5, 6]])   # 2-D (matriz)

# Arrays especiais
zeros  = np.zeros((3, 4))   # zeros 3×4
uns    = np.ones((2, 3))    # uns 2×3
ident  = np.eye(3)          # identidade 3×3
range_ = np.arange(0, 10, 2)  # [0 2 4 6 8]
linsp  = np.linspace(0, 1, 5) # [0.00 0.25 0.50 0.75 1.00]
aleats = np.random.rand(3, 3) # 3×3 com valores aleatórios em [0, 1)

# Informações
print(M.shape)   # (2, 3)
print(M.ndim)    # 2
print(M.dtype)   # int64
print(M.size)    # 6
````

### 4.3 Indexação e Fatiamento

````python
A = np.array([[10, 20, 30],
              [40, 50, 60],
              [70, 80, 90]])

# Elemento individual
print(A[0, 0])  # 10
print(A[2, 1])  # 80

# Linha / coluna completa
print(A[1, :])  # [40 50 60]
print(A[:, 2])  # [30 60 90]

# Sub-matriz
print(A[:2, 1:])  # [[20 30] [50 60]]
````

````python
# Indexação booleana
print(A[A > 50])  # [60 70 80 90]
A[A < 30] = 0     # zera elementos menores que 30
````

### 4.4 Operações Matemáticas

O NumPy aplica operações elemento a elemento automaticamente (broadcasting):

````python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

# Operações elemento a elemento
print(a + b)     # [5 7 9]
print(a * b)     # [4 10 18]
print(a ** 2)    # [1 4 9]
print(a + 10)    # [11 12 13]  — broadcasting escalar

# Funções matemáticas
print(np.sqrt(a))     # [1. 1.41 1.73]
print(np.exp(a))      # [2.72 7.39 20.09]
print(np.log(b))      # [1.39 1.61 1.79]

# Produto matricial
M1 = np.array([[1, 2], [3, 4]])
M2 = np.array([[5, 6], [7, 8]])
print(M1 @ M2)    # [[19 22] [43 50]]
````

### 4.5 Estatísticas e Agregações

````python
dados = np.array([23, 45, 12, 67, 34, 89, 56, 11, 78, 43])

print("Média  :", np.mean(dados))    # 45.8
print("Mediana:", np.median(dados))  # 44.0
print("Desvio :", np.std(dados))     # 25.85
print("Mínimo :", np.min(dados))     # 11
print("Máximo :", np.max(dados))     # 89
print("Soma   :", np.sum(dados))     # 458

# Ao longo de eixos (2-D)
M = np.array([[1, 2, 3], [4, 5, 6]])
print(np.sum(M, axis=0))  # [5 7 9]  — soma por coluna
print(np.sum(M, axis=1))  # [6 15]   — soma por linha
````

### 4.7 Exemplo Prático - Análise de Vendas Mensais

Problema: analisar as vendas de três produtos ao longo de seis meses.

````python
import numpy as np

# Linhas: produtos (Notebook, Mouse, Teclado)
# Colunas: meses (Jan a Jun)
vendas = np.array([
    [120, 135, 148, 162, 175, 190],
    [340, 298, 312, 400, 390, 420],
    [ 88,  95, 102, 115,  97, 108],
])

produtos = ["Notebook", "Mouse", "Teclado"]
meses    = ["Jan", "Fev", "Mar", "Abr", "Mai", "Jun"]

# Total por produto (soma em linha)
total_produto = vendas.sum(axis=1)
for prod, total in zip(produtos, total_produto):
    print(f"{prod:<10}: {total} unidades")

# Média mensal geral
media_mes = vendas.mean(axis=0)
for mes, med in zip(meses, media_mes):
    print(f"{mes}: {med:.1f} unidades/produto")

# Melhor mês (maior total geral)
total_mes  = vendas.sum(axis=0)
melhor_mes = meses[np.argmax(total_mes)]
print(f"\nMelhor mês: {melhor_mes} ({total_mes.max()} unidades)")

# Percentual de crescimento Jan → Jun
crescimento = (vendas[:, -1] - vendas[:, 0]) / vendas[:, 0] * 100
for prod, cresc in zip(produtos, crescimento):
    print(f"{prod:<10}: {cresc:+.1f}%")
````

### 5.2 Series - Array 1-D Rotulado

Uma Series é como uma coluna de planilha: possui valores e um índice (rótulo) para cada posição.

````python
# Criando Series
notas = pd.Series([8.5, 7.0, 9.2, 6.3],
                  index=["Ana", "Bruno", "Carla", "Diego"])

# Acesso por rótulo e por posição
print(notas["Ana"])    # 8.5
print(notas.iloc[2])  # 9.2  (posição inteira)

# Operações vetorizadas
print(notas * 10)      # multiplica todos por 10
print(notas[notas >= 7.0])  # filtra aprovados

# Estatísticas rápidas
print(notas.mean())   # média
print(notas.describe()) # resumo estatístico completo
````

💡 Índice personalizado: O índice de uma Series pode ser texto, datas ou qualquer tipo imutável - isso facilita buscas e joins sem precisar lembrar a posição numérica.

### 5.3 DataFrame - Tabela 2-D

Um DataFrame é uma tabela onde cada coluna é uma Series que compartilha o mesmo índice.

````python
# Criando DataFrame a partir de dicionário
dados = {
    "nome"    : ["Ana", "Bruno", "Carla", "Diego"],
    "idade"   : [22, 25, 21, 28],
    "nota"    : [8.5, 7.0, 9.2, 6.3],
    "aprovado": [True, True, True, False],
}
df = pd.DataFrame(dados)

# Visualizando
print(df)          # tabela completa
````

````python
# Selecionar uma coluna → Series
print(df["nome"])

# Selecionar múltiplas colunas → DataFrame
print(df[["nome", "nota"]])

# loc — seleção por rótulo de linha e coluna
print(df.loc[0, "nome"])       # "Ana"
print(df.loc[1:2, ["nome","nota"]])  # linhas 1-2, colunas nome e nota

# iloc — seleção por posição inteira
print(df.iloc[0, 0])           # "Ana"
print(df.iloc[:2, :3])         # primeiras 2 linhas, 3 colunas

# Filtro booleano
aprovados = df[df["aprovado"] == True]
acima8    = df[df["nota"] >= 8.0]
jovens_ap = df[(df["idade"] < 25) & (df["aprovado"])]
````

### 5.8 Exemplo Prático - Análise de Desempenho de Turma

Cenário: o professor recebe um CSV com notas de três bimestres e precisa calcular a média final, o conceito e gerar um relatório.

````python
import pandas as pd

# Simula um CSV já lido
df = pd.DataFrame({
    "aluno"  : ["Ana","Bruno","Carla","Diego","Elena"],
    "b1"     : [8.0, 6.5, 9.5, 5.0, 7.0],
    "b2"     : [7.5, 7.0, 8.5, 6.0, 8.0],
    "b3"     : [9.0, 5.5, 9.0, 4.5, 8.5],
})

# Calcular média final (pesos iguais)
df["media"] = df[["b1","b2","b3"]].mean(axis=1).round(2)

# Atribuir conceito
def conceito(m):
    if m >= 9.0: return "A"
    if m >= 7.0: return "B"
    if m >= 6.0: return "C"
    return "D (Reprovado)"

df["conceito"] = df["media"].apply(conceito)

# Relatório resumido
print(df[["aluno","media","conceito"]].to_string(index=False))
print()
print(f"Média da turma : {df["media"].mean():.2f}")
print(f"Aprovados      : {(df["media"] >= 6).sum()} / {len(df)}")
````

#### Anaconda Distribution

Um pacote completo que já vem com Python, Jupyter, e centenas de bibliotecas científicas pré-instaladas (NumPy, Pandas, Matplotlib, scikit-learn etc.), além do **Anaconda Navigator** (interface gráfica) e do gerenciador de ambientes **conda**.

```bash

```

- **Vantagem:** leve, isola dependências entre projetos diferentes, evita conflitos de versão.
- **Desvantagem:** exige instalar cada biblioteca manualmente.

### 3. Capítulo 3 — Listas e Dicionários em Python

Duas estruturas de dados cobrem a maior parte das necessidades de um cientista de dados antes mesmo de conhecer o Pandas: **listas** e **dicionários**.

#### Listas

Uma **lista** é uma coleção ordenada e mutável de itens, criada com colchetes:

````python
notas = [512.3, 610.7, 645.1, 470.5]
notas.append(598.4)          # adiciona ao final
print(notas[0])              # 512.3 — acesso por índice (começa em 0)
print(notas[-1])             # 598.4 — índice negativo acessa do final
print(len(notas))            # 5 — quantidade de itens
maiores = [n for n in notas if n > 550]   # list comprehension: filtragem
````

**List comprehensions** (a última linha) são uma forma concisa e "pythônica" de criar listas a partir de outras, substituindo um laço `for` explícito.

#### Dicionários

Um **dicionário** guarda pares chave-valor, criado com chaves. É a estrutura ideal para representar um registro (como uma linha de uma tabela):

````python
escola = {"nome": "Escola A", "tipo": "Estadual", "media": 512.3}
print(escola["nome"])         # "Escola A" — acesso por chave
escola["participantes"] = 180 # adiciona uma nova chave
print(escola.get("uf", "RN")) # acesso seguro, com valor padrão se a chave não existir
````

#### Combinando as duas: lista de dicionários

Uma **lista de dicionários** é a forma mais natural de representar uma pequena tabela em Python puro — cada dicionário é uma linha (registro), e a lista reúne todas as linhas. É exatamente essa estrutura que usaremos no exemplo prático desta aula, e que será substituída por um `DataFrame` do Pandas a partir da Aula 04.

````python
escolas = [
    {"nome": "Escola A", "media": 512.3},
    {"nome": "Escola B", "media": 610.7},
]
for e in escolas:
    print(e["nome"], e["media"])
````

### 4. Capítulo 4 — Funções em Python

Uma **função** encapsula um bloco de código reutilizável, evitando repetição e organizando o raciocínio em passos nomeados — essencial à medida que os scripts de análise crescem.

#### Definindo uma função

````python
def media_geral(lista_escolas):
    """Calcula a média das médias das escolas."""
    total = 0
    for escola in lista_escolas:
        total += escola["media"]
    return total / len(lista_escolas)

resultado = media_geral(escolas)   # chamando a função
````

- **def** inicia a definição; o nome deve descrever o que a função faz.
- **Parâmetros** (aqui, `lista_escolas`) são as entradas da função.
- **Docstring** (texto entre `"""`) documenta o propósito — boa prática sempre.
- **return** devolve o resultado; sem ele, a função retorna `None`.

#### Por que usar funções em análise de dados?

- **Reuso:** a mesma lógica de cálculo pode ser aplicada a diferentes conjuntos de dados sem copiar e colar código.
- **Testabilidade:** é mais fácil verificar se uma função pequena está correta do que validar um script inteiro de uma vez.
- **Legibilidade:** nomes de função bem escolhidos (`filtrar_por_tipo`, `escola_com_maior_media`) tornam o código autoexplicativo.

#### Funções com múltiplos parâmetros e valores padrão

````python
def filtrar_por_tipo(lista_escolas, tipo="Estadual"):
    return [e for e in lista_escolas if e["tipo"] == tipo]

filtrar_por_tipo(escolas)              # usa o padrão "Estadual"
filtrar_por_tipo(escolas, "Privada")   # sobrescreve o padrão
````

No exemplo prático desta aula, você verá três funções trabalhando juntas sobre a mesma lista de dicionários: uma calcula uma média, outra filtra registros, e outra encontra um extremo — o mesmo tipo de composição que reaparecerá, com Pandas, a partir da Aula 04.

### 5. Capítulo 5 — Introdução ao NumPy: Arrays e Operações Vetorizadas

O **NumPy** (Numerical Python) é a base de praticamente todo o ecossistema científico em Python — Pandas, Matplotlib e scikit-learn são construídos sobre ele. Sua estrutura central é o **array**.

#### Array x lista: qual a diferença?

Um array do NumPy parece uma lista, mas é fundamentalmente diferente: armazena elementos do **mesmo tipo** em um bloco contíguo de memória, o que permite operações matemáticas muito mais rápidas e diretas.

````python
import numpy as np

medias = np.array([512.3, 610.7, 645.1, 470.5])
print(medias.mean())    # 559.65 — média, sem escrever um loop
print(medias.std())     # desvio padrão
print(medias.max())     # valor máximo
print(medias * 1.1)     # multiplica CADA elemento por 1.1, de uma vez
````

#### Operações vetorizadas

Isso é **vetorização**: aplicar uma operação a todos os elementos de um array simultaneamente, sem escrever um laço `for` explícito. O código fica mais curto, mais legível e — como veremos no exemplo prático — muito mais rápido.
